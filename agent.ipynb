```python
import os
import openai

apikey = os.getenv("TIGER_AI_GATEWAY_API_KEY")  # Read from environment
base_url = "https://api.ai-gateway.tigeranalytics.com"
model = "gpt-5-nano"
user_query = "What is the capital of France?"   # The text we want to ask Model.

client = openai.OpenAI(api_key=apikey, base_url=base_url)
response = client.chat.completions.create(
    model=model,
    messages=[{"role": "user", "content": user_query}],
)
print(response)
```
